```python
import cv2
import numpy as np
import math
import time


OBJECT_RECT = (400, 150, 150, 150)  # (x, y, w, h)
WARNING_DIST = 150
DANGER_DIST = 40
FONT = cv2.FONT_HERSHEY_SIMPLEX

class HandTracker:
    def __init__(self):
        self.cap = cv2.VideoCapture(0)
        self.is_calibrated = False

        # Default HSV ranges (fallback)
        self.lower_skin = np.array([0, 20, 70], dtype=np.uint8)
        self.upper_skin = np.array([20, 255, 255], dtype=np.uint8)

        self.cal_rect_size = 60

    def calibrate(self, frame):
        h, w, _ = frame.shape
        cx, cy = w // 2, h // 2
        roi = frame[cy - self.cal_rect_size//2 : cy + self.cal_rect_size//2,
                    cx - self.cal_rect_size//2 : cx + self.cal_rect_size//2]
        hsv_roi = cv2.cvtColor(roi, cv2.COLOR_BGR2HSV)
        mean_color = np.mean(hsv_roi, axis=(0, 1))

        hue_offset = 15
        sat_offset = 50
        val_offset = 50

        self.lower_skin = np.array([
            max(0, mean_color[0] - hue_offset),
            max(20, mean_color[1] - sat_offset),
            max(20, mean_color[2] - val_offset)
        ], dtype=np.uint8)

        self.upper_skin = np.array([
            min(180, mean_color[0] + hue_offset),
            min(255, mean_color[1] + sat_offset),
            min(255, mean_color[2] + val_offset)
        ], dtype=np.uint8)

        print(f"Calibrated! Lower: {self.lower_skin}, Upper: {self.upper_skin}")
        self.is_calibrated = True

    def get_hand_contour(self, frame):
        hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)
        mask = cv2.inRange(hsv, self.lower_skin, self.upper_skin)

        # morphological ops
        kernel = np.ones((5,5), np.uint8)
        mask = cv2.erode(mask, kernel, iterations=1)
        mask = cv2.dilate(mask, kernel, iterations=2)
        mask = cv2.GaussianBlur(mask, (5, 5), 0)

        contours, _ = cv2.findContours(mask, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)
        if contours:
            # pick largest
            c = max(contours, key=cv2.contourArea)
            if cv2.contourArea(c) > 1000:
                return c
        return None

    def closest_point_on_rect(self, px, py, rect):
        rx, ry, rw, rh = rect
        # clamp point to rectangle
        cx = min(max(px, rx), rx + rw)
        cy = min(max(py, ry), ry + rh)
        return (int(cx), int(cy))

    def calculate_distance_to_rect(self, point, rect):
        px, py = point
        rx, ry, rw, rh = rect
        dx = max(rx - px, 0, px - (rx + rw))
        dy = max(ry - py, 0, py - (ry + rh))
        return math.sqrt(dx*dx + dy*dy)

    def draw_text_with_outline(self, img, text, org, font, scale, color, thickness):
        # draw black outline then colored text (for visibility)
        cv2.putText(img, text, org, font, scale, (0,0,0), thickness+4, cv2.LINE_AA)
        cv2.putText(img, text, org, font, scale, color, thickness, cv2.LINE_AA)

    def run(self):
        print("Starting Hand Tracker...")
        print("1. Place hand in the center green box.")
        print("2. Press 'c' to Calibrate skin tone.")
        print("3. Press 'q' to Quit.")

        # small FPS measurement
        last_time = time.time()
        fps = 0.0

        while True:
            ret, frame = self.cap.read()
            if not ret:
                break

            frame = cv2.flip(frame, 1)
            h, w, _ = frame.shape
            display_frame = frame.copy()

            if not self.is_calibrated:
                cx, cy = w // 2, h // 2
                # Draw calibration box (black outline + green inner)
                box_tl = (cx - self.cal_rect_size//2, cy - self.cal_rect_size//2)
                box_br = (cx + self.cal_rect_size//2, cy + self.cal_rect_size//2)
                cv2.rectangle(display_frame, box_tl, box_br, (0,0,0), 6)
                cv2.rectangle(display_frame, box_tl, box_br, (0,255,0), 2)
                cv2.putText(display_frame, "Place hand in box & press 'c'", (cx - 170, cy - 60),
                            FONT, 0.7, (0,255,0), 2)
            else:
                contour = self.get_hand_contour(frame)
                state = "SAFE"
                color = (0, 255, 0)
                distance = None

                if contour is not None:
                    # draw black thick contour underneath for contrast
                    cv2.drawContours(display_frame, [contour], -1, (0,0,0), 8)
                    # then colored contour
                    cv2.drawContours(display_frame, [contour], -1, (255, 255, 0), 2)

                    # convex hull to find fingertip-like point
                    hull = cv2.convexHull(contour)
                    # pick topmost point from hull (min y)
                    topmost = tuple(hull[hull[:,:,1].argmin()][0])

                    # draw black dot under fingertip then colored dot
                    cv2.circle(display_frame, topmost, 12, (0,0,0), -1)   # black base
                    cv2.circle(display_frame, topmost, 7, (0,0,255), -1)  # red fingertip

                    # compute distance and closest point on rectangle
                    distance = self.calculate_distance_to_rect(topmost, OBJECT_RECT)
                    closest = self.closest_point_on_rect(topmost[0], topmost[1], OBJECT_RECT)

                    # determine state
                    if distance < DANGER_DIST:
                        state = "DANGER"
                        color = (0, 0, 255)
                    elif distance < WARNING_DIST:
                        state = "WARNING"
                        color = (0, 255, 255)
                    else:
                        state = "SAFE"
                        color = (0, 255, 0)

                    # draw strong black line (thick) then colored thin line from fingertip -> closest point on rect
                    cv2.line(display_frame, topmost, closest, (0,0,0), 8, cv2.LINE_AA)   # black backbone
                    cv2.line(display_frame, topmost, closest, color, 3, cv2.LINE_AA)     # colored overlay

                    # draw small black circle under the closest point and colored circle
                    cv2.circle(display_frame, closest, 10, (0,0,0), -1)
                    cv2.circle(display_frame, closest, 6, color, -1)

                # Draw object rectangle with black base -> colored outline and subtle fill
                ox, oy, ow, oh = OBJECT_RECT
                # black thick border
                cv2.rectangle(display_frame, (ox, oy), (ox+ow, oy+oh), (0,0,0), 8)
                # colored thinner border
                cv2.rectangle(display_frame, (ox, oy), (ox+ow, oy+oh), color, 3)

                # subtle inner fill (only slightly visible)
                overlay = display_frame.copy()
                if state == "DANGER":
                    # pulsing fill effect using time
                    alpha = 0.35 + 0.15 * math.sin(time.time()*8)
                    fill_color = (0, 0, 255)
                    cv2.rectangle(overlay, (ox+3, oy+3), (ox+ow-3, oy+oh-3), fill_color, -1)
                    cv2.addWeighted(overlay, alpha, display_frame, 1 - alpha, 0, display_frame)
                    # big DANGER text with black outline
                    self.draw_text_with_outline(display_frame, "DANGER DANGER", (40, h - 40), FONT, 1.8, (0,0,255), 3)
                elif state == "WARNING":
                    alpha = 0.12
                    fill_color = (0,255,255)
                    cv2.rectangle(overlay, (ox+3, oy+3), (ox+ow-3, oy+oh-3), fill_color, -1)
                    cv2.addWeighted(overlay, alpha, display_frame, 1 - alpha, 0, display_frame)
                else:
                    alpha = 0.06
                    fill_color = (0,255,0)
                    cv2.rectangle(overlay, (ox+3, oy+3), (ox+ow-3, oy+oh-3), fill_color, -1)
                    cv2.addWeighted(overlay, alpha, display_frame, 1 - alpha, 0, display_frame)

                # UI overlay box with small black background for readability
                cv2.rectangle(display_frame, (10, 10), (360, 120), (0,0,0), -1)
                # show state with colored text (draw text with black outline)
                self.draw_text_with_outline(display_frame, f"State: {state}", (20, 50), FONT, 0.9, color, 2)
                dist_text = f"Dist: {int(distance)}" if distance is not None else "Dist: --"
                cv2.putText(display_frame, dist_text, (20, 90), FONT, 0.8, (220,220,220), 1, cv2.LINE_AA)

            # FPS calc
            now = time.time()
            fps = 0.9 * fps + 0.1 * (1.0 / (now - last_time)) if (now - last_time) > 0 else fps
            last_time = now
            cv2.putText(display_frame, f"FPS: {fps:.1f}", (display_frame.shape[1]-140, 30), FONT, 0.6, (200,200,200), 1)

            cv2.imshow("Arvyax Assignment - Hand Guard (Improved Viz)", display_frame)

            key = cv2.waitKey(1) & 0xFF
            if key == ord('q'):
                break
            elif key == ord('c'):
                self.calibrate(frame)

        self.cap.release()
        cv2.destroyAllWindows()

if __name__ == "__main__":
    app = HandTracker()
    # init fps variable used in draw loop
    last_time = time.time()
    fps = 0.0
    app.run()
```
